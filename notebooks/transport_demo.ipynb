# Monte Carlo photon transport: a small reproducible example
Install this repository with `python -m pip install -e .` before opening this notebook.
The historical notebook is preserved separately. This example uses an educational, unvalidated coefficient table. Values represent deposited energy, not absorbed dose.


```python
from stochastic_photons.transport import simulate_ensemble
result = simulate_ensemble(histories=3, photons=1000, grid_size=12, seed=42)
print('Mean deposited energy per primary (MeV):', result['mean_mev_per_primary'].sum())
for balance in result['energy_balances']:
    accounted = sum(balance[k] for k in ('deposited_mev', 'escaped_mev', 'unresolved_mev'))
    assert abs(accounted - balance['source_mev']) < 1e-10
```

The standard error is computed across independent batch means with the sample variance (ddof=1). A zero voxel tally does not establish that the underlying expectation is zero. Three small batches illustrate the API; they do not establish convergence.


```python
from pathlib import Path
from stochastic_photons.plotting import plot_transport
destination = Path('output/notebook/transport.png')
destination.parent.mkdir(parents=True, exist_ok=True)
plot_transport(result, 10.0, destination)
print('Saved', destination)
```
